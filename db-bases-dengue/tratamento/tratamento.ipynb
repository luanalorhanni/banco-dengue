```python
import pandas as pd
import sqlite3 as sql
#import numpy as np
```

```python
dengbr21 = pd.read_csv('dengbr21.csv',delimiter=';')
dengbr22 = pd.read_csv('dengbr22.csv',delimiter=',')
dengbr19 = pd.read_csv('dengbr19.csv',delimiter=',')
dengbr18 = pd.read_csv('dengbr18.csv',delimiter=',')
dengbr17 = pd.read_csv('dengbr17.csv',delimiter=',')
dengbr16 = pd.read_csv('dengbr16.csv',delimiter=',')
```

```python
unificada = [dengbr16,dengbr17,dengbr18,dengbr19,dengbr21,dengbr22]

for planilha in unificada:
    count_colunas = len(planilha.columns)
    print(count_colunas)
```

```python
banco_dengue = pd.concat(unificada, ignore_index=True)
banco_dengue.head()
```

```python
banco_dengue.info()
```

```python
banco_dengue.describe()
```

```python
banco_dengue.isnull().sum()
```

```python
banco_dengue = banco_dengue.fillna('Não Informado')
banco_dengue.isnull().sum()
```

```python
colunas = banco_dengue.columns
colunas_dt = [col for col in colunas if "DT_" in col] #List comprehension
colunas_dt
```

```python
for coluna in colunas_dt:
    banco_dengue[coluna] = pd.to_datetime(banco_dengue[coluna], errors='coerce')
    print(banco_dengue[coluna].count())
```

```python
colunas = banco_dengue.columns

for coluna in colunas:
    tipo_coluna = banco_dengue[coluna].dtype
    print(f'{coluna}: {tipo_coluna}')
```

```python
# Calcular a diferença entre as dias das datas de notificação e as datas dos sintomas
banco_dengue['diferenca_dias'] = (banco_dengue['DT_NOTIFIC'] - banco_dengue['DT_SIN_PRI']).dt.days

# Ordenar o DataFrame pela coluna 'diferenca_dias' em ordem decrescente
dengue_ordenado = banco_dengue.sort_values(by='diferenca_dias', ascending=False)
resultado_dias = dengue_ordenado[['DT_NOTIFIC', 'DT_SIN_PRI', 'diferenca_dias']]
resultado_dias.head(30)
```
